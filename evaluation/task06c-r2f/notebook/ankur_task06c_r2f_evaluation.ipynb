# Ankur Task 06C-R2F final assessment evaluation

Provider-free reconciliation of the single frozen public-safe run. Historical Task 06, Task 06C, R1, R2, and R2E remain separate.

```python
import json
from pathlib import Path
root = Path('evaluation/task06c-r2f/run-1')
metrics = json.loads((root / 'exports' / 'aggregate-metrics.json').read_text(encoding='utf-8'))
operations = json.loads((root / 'records' / 'public' / 'provider-operations.json').read_text(encoding='utf-8'))
attempts = json.loads((root / 'records' / 'public' / 'provider-attempts.json').read_text(encoding='utf-8'))
questions = json.loads((root / 'records' / 'public' / 'question-records.json').read_text(encoding='utf-8'))
written = json.loads((root / 'records' / 'public' / 'written-grading-records.json').read_text(encoding='utf-8'))
diagnostics = json.loads((root / 'records' / 'public' / 'semantic-diagnostics.json').read_text(encoding='utf-8'))
decision = json.loads((root / 'RUN_DECISION.json').read_text(encoding='utf-8'))
print('Loaded public-safe Task 06C-R2F run 1')
```

```python
assert decision['classification'] == 'KEEP'
assert decision['technicalGate'] == 'failed'
assert metrics['reliability']['totalOperations'] == len(operations) == 45
assert len(attempts) == 60
assert metrics['questions']['total'] == len(questions) == 42
assert metrics['written']['total'] == len(written) == 18
assert metrics['questions']['deterministicGrounding']['percentage'] == 100
assert metrics['questions']['deterministicKeyValidity']['percentage'] == 100
assert all(metrics['reliability'][key] == 0 for key in ['groundingFailures', 'quoteFailures', 'conceptFailures', 'reconciliationFailures'])
print('R2F final validity: 33/45; questions: 42; written: 18')
print('Provider-attempt rows:', len(attempts))
print('Sanitized semantic diagnostics:', len(diagnostics))
print('Fixed-denominator and artifact reconciliation: PASSED')
```

```python
gates = {
    'final_valid_43_of_45': metrics['reliability']['finalValid']['count'] >= 43,
    'questions_at_least_30': len(questions) >= 30,
    'written_at_least_10': len(written) >= 10,
    'grounding_100_percent': metrics['questions']['deterministicGrounding']['percentage'] == 100,
    'mcq_key_100_percent': metrics['questions']['deterministicKeyValidity']['percentage'] == 100,
    'invalid_rubrics_zero': metrics['reliability']['reconciliationFailures'] == 0,
}
for gate, passed in gates.items():
    print(gate, 'PASS' if passed else 'FAIL')
assert gates['final_valid_43_of_45'] is False
assert all(value for key, value in gates.items() if key != 'final_valid_43_of_45')
print('Human metrics: pending; reviewer packets: not generated')
print('Task 07 authorized: NO')
```
